## Infrared (IR) Sensors

IR sensors detect **reflected infrared light** to sense color or proximity. A [comparator](https://www.electronics-tutorials.ws/opamp/op-amp-comparator.html) is a circuit which compares a varying input voltage (Vin) to a steady reference voltage (V_ref). The [LM393 comparator](https://www.ti.com/product/LM393) on an IR sensor chip outputs a high or low signal based on light reflection strength.

- **Comparator Circuit**: The comparator outputs a digital signal (Vcc or 0V) based on detected IR light.
- Useful for **line following** or **edge detection**.

## Integrating HC-SR04 Ultrasonic Sensor

The **HC-SR04** is an ultrasonic ToF sensor with four pins: **Vcc**, **Gnd**, **Trig**, and **Echo**. 

1. Connect the Vcc and Gnd to the 5V bus

![The HC-SR04 and jumper cables plugged into the mini breadboard](4.6.png)

2. Echo to pin A29 and Trig to pin A30 on the Arduino.

![Jumpers from Echo and Trig plugged into A29 and A30 on the breadboard](4.7.png) 

The setup for ultrasonic measurement allows us to return the measured distance based on echo time after triggering a pulse.
:

```cpp
// declare echo and trig pins for ultrasonic sensor 
const int echo = 11; // Echo -> D11
const int trig = 12; // Trig -> D12

 //--------------------setup sensor pins-------------------- 
void setup() {
    pinMode(echo, INPUT);
    pinMode(trig, OUTPUT);
}

 // send out an ultrasonic pulse thats 10ms long 
float get_distance() {
    digitalWrite(trig, HIGH);
    delayMicroseconds(10);
    digitalWrite(trig, LOW);
    
    // use pulseIn function to see how long it takes for the pulse to return to the  sensor 
    long echo_time = pulseIn(echo, HIGH); 
    // calculate distance using formula from ToF sensor section 
    calculated_distance = (echo_time / 2) / (346 * 10);  
    //*10 is to scale units appropriately since pulseIn returns in microseconds 

    return calculated_distance  // distance in meters
}
```

## Setting Up the IR Light Sensor

The infrared light sensor is even simpler than the ultrasonic time-of-flight sensor to use. First,  use your fingers to bend the IR LED and Phototransistor to be perpendicular to the sensor board.

![The IR sensor with pins, LED, and phototransistor bent perpendicular to the board](4.8.png)

The **IR sensor** uses three pins:
- **VCC** and **GND** for power, which we connect to the 5V bus.
- **OUT** pin, which we connect to **J30** on the breadboard.

![The IR sensor and jumpers plugged into the mini breadboard](4.9.png)

Add these declarations to your Arduino code:
```cpp
// Declare IR sensor pin
const int ir = 13;  // OUT -> D13

void setup() {
    pinMode(ir, INPUT);
}
```

We can use digitalRead to check the value of the IR sensor output. If the IR sensor sees a white line, it’ll set the output pin to 0V and digitalRead will return a value of LOW. Otherwise, the IR sensor output pin will be at 5V and digitalRead will return HIGH. Conveniently, [the value of HIGH and LOW can be substituted for TRUE and FALSE (due to how these variables are represented in binary code)](https://forum.arduino.cc/t/interchanging-high-low-with-true-false/146611). Here’s the function:

```cpp
bool get_line() {
    return !digitalRead(ir);  // Returns TRUE if line is detected
}
```

## Integrating Motor Encoders

The **quadrature encoders** on the rover’s motors track wheel rotation and help estimate distance traveled, also known as **odometry**. Integrating motor-driven quadrature encoders can be challenging because the speed at which the encoder pins rise and fall may be faster than the speed at which a microcontroller loops through a
program, which can cause the microcontroller to miss counts. There are [hardware-based solutions](https://forum.digikey.com/t/design-a-robust-quadrature-encoder-isr-based-program-for-arduino-motor-control/39913) to this problem, and it can be an educational problem to solve on your own.

## Calculating Odometry in get_odom()

To track the rover’s movement, we use encoder counts to calculate the wheel’s angular position and the distance traveled (odometry).

1. **Get Encoder Counts**: We use the `getEncoderCount()` method to read encoder counts.
2. **Convert Counts to Wheel Position**: Calculate the angular wheel position based on encoder counts.
3. **Calculate Distance**: Using forward kinematics, find the distance traveled by the robot.

With some [dimensional analysis](https://www.chem.tamu.edu/class/fyp/mathrev/mr-da.html), we find that for each 2*pi radians travelled, we’d register 3575.04 encoder counts.

```cpp
long left_encoder_count = left_encoder.getEncoderCount();
long right_encoder_count = right_encoder.getEncoderCount();

// Calculate wheel position
float left_wheel_pos = left_encoder_count * ((2 * 3.14) / 3575.04); // 3575.04 is the counts per revolution.
float right_wheel_pos = right_encoder_count * ((2 * 3.14) / 3575.04);

// Calculate odometry
float odom = (r / 2) * (left_wheel_pos + right_wheel_pos);
return odom;
```

If you notice incorrect readings (e.g., negative values), try swapping the yellow and white encoder wires.

## Defining FSM States Using Enum

To represent each state, we’ll create an enumeration (enum). Using enum makes our code readable by allowing us to work with descriptive names rather than numbers.

### Code Example:
```cpp
// Set up FSM states
enum STATE {follow_right, turn_right, follow_left, turn_left, stop};
STATE last_state;
STATE current_state = follow_right;
STATE next_state = current_state;
```

Explanation:
- last_state: Tracks the previous state.
- current_state: The FSM’s active state.
- next_state: Updates to the next state after evaluating events.

## Creating FSM Transitions Using switch:case

The **switch:case** structure evaluates **current_state** in each loop iteration, moving the program to the relevant state block. Each block has:
1. **Event Checkers**: Use if statements to check conditions for state transitions.
2. **Actions**: Commands executed in the current state.

### Code Example:
```cpp
switch (current_state) {
    case follow_right:
        if (get_odom() > 5) {
            next_state = stop;
            break;
        }
        if (get_distance() < 0.1) {
            next_state = turn_right;
            break;
        }
        // Actions for follow_right
        if (get_line()) motor_controller(0.1, -1);  // Turn right
        else motor_controller(0.1, 1);  // Turn left
        break;
    
    case stop:
        motor_controller(0, 0);
        delay(3000);
        next_state = (last_state == follow_right) ? turn_right : turn_left;
        break;
}
```

## Debouncing for Stable Transitions

Debouncing avoids immediate transitions that may skip over necessary actions, especially when switching directions. In **turn_right**, we add a short delay to ensure the line isn’t detected prematurely.

### Code Example:
```cpp
case turn_right:
    motor_controller(0, -3);  // Turn in place
    if (last_state != turn_right) delay(500);  // Debounce to clear line
    if (get_line()) next_state = follow_left;
    break;
```
Adding delay after detecting the transition stabilizes the FSM, allowing accurate direction detection.



## Updating States

After evaluating events and actions in the switch:case, the FSM must **update states** for the next loop.

### Code Example:
```cpp
last_state = current_state;
current_state = next_state;
```
By updating these variables, we ensure the FSM enters the correct state in the next loop iteration.
