## Power Concepts

Before we design the rover’s power circuit, let’s review some important terms:

| **Term**             | **Unit**     | **Symbol** |
|----------------------|--------------|------------|
| Voltage (Electric Potential) | Volts (V)  | V          |
| Current             | Amps (A)     | I          |
| Power               | Watts (W)    | P          |

There is a lot to learn about circuits, and many people through the years have done a much better
[job explaining circuits](https://ataridogdaze.com/science/hydraulic/index.html) than we have time for here.

The relationship between these terms is summarized by the equation:

$$ P = V \times I $$

### Important Notes:
1. **Constant Voltage, Varying Current**: Most components require a constant voltage, but their current draw can vary depending on the device. For example, a microcontroller might draw more current than a sensor, even if both run at 5V.
2. **Maximum Ratings**: A power supply’s rating tells us the maximum power or current it can safely deliver. Exceeding this could lead to overheating or damage.
3. **Safe Design**: Always match components, batteries, and wires to ensure they work safely together. Mismatched power components can lead to malfunction or dangerous situations.

## Battery Selection for the Rover

To power our rover, we need a battery that can supply **1.55A** at **12V**. A **3-cell (3S) LiPo battery** is a good choice because it provides approximately 12V.

### LiPo Battery Specs:

When selecting a LiPo battery, here are three important specifications to keep in mind:

1. **S-number (Cell Count)**: The number of cells in the battery determines the voltage. Each LiPo cell provides about **4V**, so a **3S battery** gives us around **12V**.
2. **Capacity (mAh)**: This indicates how long the battery can provide current. For example, a **500 mAh** battery can supply **1.55A** for approximately **20 minutes**.
3. **Discharge Rating (C-number)**: The C-number tells us how much current the battery can safely supply. You can calculate this by multiplying the battery capacity by the C-number. For example, a **500 mAh, 35C** battery can supply a maximum of **17.5A**—far more than what we need.

### Safety Notes for LiPo Batteries:
LiPo batteries are powerful but must be handled with care. Here are a few safety guidelines to follow:
- Do not let the battery charge get too low, as it will degrade the battery and could make it unsafe to use.
- If the battery becomes puffy or damaged, dispose of it properly.
- Always charge LiPo batteries using a proper LiPo charger in a fire-safe area, and have safety measures in place.

Let’s calculate the runtime of our selected battery.


## Calculating Battery Runtime

To estimate how long our battery will last, we use the following formula:

$$ \text{Runtime (hours)} = \frac{\text{Battery Capacity (mAh)}}{\text{Current Draw (mA)}} $$

For example, if we have a **500 mAh battery** and our current draw is **1550 mA**, the runtime will be:

$$ \frac{500}{1550} = 0.32 \text{ hours} = 20 \text{ minutes} $$

This calculation tells us the battery will power the rover for about **20 minutes**.

## LiPo Battery Safety

LiPo batteries are excellent for robotics due to their ability to store and release energy quickly, but they must be handled with respect. Here are some key safety points:

1. **Battery Charge**: Never let the battery charge get too low, as it may become unsafe to use.
2. **Physical Condition**: If the battery becomes puffy, squishy, or damaged, dispose of it safely in a battery disposal bin.
3. **Charging**: Always use a proper LiPo charger, charge in a fire-resistant area, and monitor the charging process.

### Charging Safety Tips:
- Have a fire blanket, sandbags, or a metal container nearby during charging.
- Never leave the battery unattended while charging.
- Avoid charging the battery in enclosed spaces where heat may build up.

By following these guidelines, you can safely use LiPo batteries in your projects.


## Other Considerations for Power Circuits

We’ve designed an effective power circuit, but there are a few additional considerations for making your circuit more usable and safe in future designs.

### Key Considerations:
1. **Wire Gauge**: Ensure that your wires can handle the current. [Wire gauge table](https://www.powerstream.com/Wire_Size.htm).
2. **Fuses**: Adding a fuse after the battery can prevent dangerous short circuits by cutting off power if the current exceeds a safe limit.
3. **Power Switches**: A convenient power switch is useful for turning the robot on and off without needing to plug and unplug the battery. This can also prevent short circuits.

## Assembling the Voltage Regulator

Let’s start by assembling the voltage regulator, which will step down the 12V supply to 5V for our microcontroller and sensors.

![assembly1](2.5.png)


### Steps:
1. Gather a **LiPo battery connector (JST Plug)** and four 10cm **jumper cables**.
2. Trim and strip the ends of the cables to expose the wire.
3. Twist two jumper wires with the battery connector wires and solder them to the **IN side** of the voltage regulator (red to +, black to -).
4. Solder the other two jumper cables to the **OUT side** of the regulator.

![assembly2](2.6.png)


Optional: Use heat shrink tubing to cover the soldered wires and keep everything tidy.

Once the voltage regulator is assembled, it should look like this:

## Placing Components on the [Breadboard](https://www.sciencebuddies.org/science-fair-projects/references/how-to-use-a-breadboard)

Now that we’ve assembled the voltage regulator and motor driver, let’s place them on the breadboard and connect them to the power buses.

### Steps:
1. Place the **motor driver** on the breadboard with the largest chip facing the front of the rover. Press the left row of pins into D6-D13 and the right row into H6-H13.
2. Connect the **output of the voltage regulator** (5V) to the right power rail of the breadboard (your 5V bus).
3. Connect the **input of the voltage regulator** (12V) to the left power rail of the breadboard (your 12V bus).
4. Use jumper cables to connect the **VIN and VCC pins** on the Arduino and motor driver to the 5V bus.
5. Connect the **GND pins** of the Arduino and motor driver to the ground rail (right-).
6. Connect the **STBY pin** on the motor driver (A10) to the 5V bus.

Your power circuit should now look like this:

![assembly6](2.9.png)



## Connecting the 12V Components

The only 12V components in the rover are the motors, and they receive 12V through the motor driver.

### Steps:
1. Use a jumper wire to connect the **12V rail** (left+) to the **VM pin** on the motor driver (J13).
2. No additional ground connection is needed because the 12V and 5V grounds are already connected via the voltage regulator.

## Connecting the Motors

Now let’s connect the motors to the motor driver.

### Steps:
1. Twist 1 red and 1 black jumper wire for each motor.
2. Connect the **left motor** by plugging the red wire into E1 and I8 and the black wire into E2 and I7.
3. Connect the **right motor** by plugging the red wire into F1 and I10 and the black wire into F2 and I9.
