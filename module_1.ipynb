## What is a Differential Drive Robot?

A differential drive robot has two driven wheels, one on the left and one on the right. These wheels control the motion of the robot. The robot can move forward, backward, and spin left and right by adjusting the speed of each wheel. Extra wheels are often added for stability and minimizing friction, called caster wheels.

### Advantages:
- Simple and nimble, capable of turning on a dime (zero-point turn).
- Effective on flat and level surfaces.

### Disadvantages:
- May struggle on uneven terrain or areas requiring better traction.

## Robot Motion and Velocity

In order to control the movement of our robot, we need to understand the relationship between the wheels’ movement and the robot’s body motion. The robot has two types of velocity:
- **Linear velocity (v):** How fast the robot moves forward or backward.
- **Angular velocity (w):** How fast the robot rotates left or right.

### Key Formula:
The robot’s wheel angular velocity (ø-dot) is related to its linear velocity (v) by the equation:

$$v = r \times \dot{ø}$$

Where:
- **r** is the radius of the wheel.
- **ø-dot** is the angular velocity of the wheel (in radians/second).
- **v** is the linear velocity of the center of the wheel.

This tells us how the turning of the wheels affects the robot’s movement.

## Kinematics of Differential Drive Robots

Now that we've covered the basics, let's explore how the wheels’ motion relates to the overall motion of the robot body.

When both wheels turn in the same direction at the same speed, the robot moves **straight forward**. When the wheels turn in opposite directions, the robot **spins in place**. 

### Special Cases:
- **Both wheels turn in the same direction at the same speed**: Robot moves straight forward.
- **Wheels turn in opposite directions**: Robot spins in place (zero-point turn).

But how can we describe the relationship between the wheel speeds and the robot's linear and angular velocities? This is where inverse and forward kinematics come into play.

### Inverse Kinematics

Inverse kinematics allows us to calculate the wheel speeds needed to achieve a desired robot velocity. If we know the desired forward velocity \(v\) and the desired angular velocity \(w\), we can calculate the angular velocities of the left (\( \dot{\phi}_L \)) and right (\( \dot{\phi}_R \)) wheels using these equations:

$$\dot{\phi}_L = \frac{v - \frac{w \cdot d}{2}}{r}$$

$$\dot{\phi}_R = \frac{v + \frac{w \cdot d}{2}}{r}$$

Where:
- \(v\) is the linear velocity of the robot.
- \(w\) is the angular velocity of the robot.
- \(r\) is the radius of the wheels.
- \(d\) is the distance between the left and right wheels.

### Forward Kinematics

Forward kinematics, on the other hand, lets us calculate the robot’s velocity based on the known wheel speeds. If we know the angular velocities of the left and right wheels $( \dot{\phi}_L )$ and $( \dot{\phi}_R) $, we can calculate the robot's linear and angular velocities using these equations:

![kinematics](kinematics.png)

$$v = \frac{r}{2} (\dot{\phi}_R + \dot{\phi}_L)$$

$$w = \frac{r}{d} (\dot{\phi}_R - \dot{\phi}_L)$$

Where:
- \(v\) is the linear velocity of the robot.
- \(w\) is the angular velocity of the robot.
- \(r\) is the radius of the wheels.
- \(d\) is the distance between the left and right wheels.

## Assembling the Rover

Once all the parts are ready, we can assemble the rover. Here are the key steps:

1. **Mount the Wheels**: Line up the flat side of the axle with the flat side of the wheel hole to prevent slipping.

- Notice the motor axles are D shaped; they have a flat side. Make sure you line this flat side of the axle up with the flat side of  the axle hole in the wheel when you slide the wheel on. This prevents the wheel from slipping  when the motor turns. Be careful as you slide the motor axle into the wheel, be gentle with the  motor so you don’t damage it.
    
    ![wheels](1.10.png)


2. **Attach the Motors**: Use motor brackets, bolts, and nuts to secure the motors to the chassis.

- If the gearbox extends out past the white motor mounts, you’ve mounted  your motors improperly.

    ![wheels2](1.11.png)


3. **Install the Caster Wheel**: Use bolts and nuts to attach the caster wheel to the front underside of the chassis.

- You may need a pair of pliers to hold the nuts still while you tighten the bolts. Only once the bolts are tight enough they won’t  come loose can you press the metal caster ball into the housing.

    ![casterwheels](1.12.png)


4. **Place the Breadboards**: Stick the breadboards onto the chassis using the sticker backing. Place the large breadboard at the back and the small one at the front.

    ![breadboard](1.13.png)



## Soldering the Motor Cable Connectors

Now let's solder the motor cable connectors. Here’s how to do it:

1. **Snap off 6 header pins** for each connector.
2. Stick the pins into a breadboard to stabilize them while soldering.
3. Solder one pin to the connector. The connector may "droop" down, so reheat the solder and adjust it with your free hand.
4. Wait for the solder to cool, then solder the rest of the pins.
5. Repeat the process for the second connector.

![breadboard2](1.14.png)

Once the connectors are soldered, we’ll place the components on the breadboard.


## Placing Components on the Breadboard

Once the motor cable connectors are soldered, we can start placing the components on the breadboard:

1. **Plug the motor cables** into the motors and connectors. Ensure the plugs are inserted in the correct direction.
2. Insert the **left motor connector** into **A1-A6** and the **right motor connector** into **J1-J6** on the large breadboard. Ensure the red wire faces forward.

Next, let’s place the **Arduino Nano** on the breadboard:

1. Make sure the **USB connector** is facing the back of the rover.
2. Press the left row of Arduino pins into **D16-D30** and the right row of pins into **H16-H30**.
