## Supply Voltage and DC Motors

Our rover uses [brushed DC motors](https://en.wikipedia.org/wiki/Brushed_DC_electric_motor) (BDC motors) which are simpler and ideal for our purpose. Unlike brushless motors (BLDC motors), which are great for high-speed applications, BDC motors are reliable, affordable, and easy to control.

We control the speed of a BDC motor by adjusting the **supply voltage**. At 12V, the motor runs at full speed. To make it run slower, we apply less voltage.

The **direction** of the motor is controlled by switching the voltage polarity across the motor terminals.

### Motor Direction
To change the direction of a BDC motor, we switch the polarity of the voltage across its terminals. The motor doesn’t care which terminal is positive or negative; it only cares that there is voltage across its terminals.

## Varying Motor Speed with PWM

To control the speed of a BDC motor, we vary the **voltage**. But, with a constant 12V battery, we can’t adjust voltage directly. Instead, we use **Pulse Width Modulation (PWM)**.

### How PWM Works:
- PWM involves turning the 12V supply on and off rapidly (hundreds of times per second).
- By changing the amount of time the voltage is **on** (pulse width) vs **off**, we create an **apparent voltage** the motor sees.

### Duty Cycle
- The **duty cycle** is the percentage of each cycle when the voltage is on. For example:
  - A 75% duty cycle means the voltage is on 75% of the time, so the motor sees 75% of 12V (9V).

The Arduino generates these high-frequency pulses, while the **motor driver** acts as a switch to control the voltage the motor receives.

## Reversing Motor Direction with an H-Bridge

To make our motors run backward, we need to **reverse the voltage** across the motor terminals. The **H-bridge circuit** in the motor driver allows this reversal without manually swapping wires.

### How an H-Bridge Works:
- An H-bridge consists of **four transistors** acting as switches.
- By activating opposite pairs of transistors, the voltage polarity changes, reversing motor direction.

![How an H-bridge circuit reverses motor direction](3.5.png)

This H-bridge setup enables smooth forward and backward control without needing to rewire.

## Arduino setup() Function

In Arduino code, the **setup()** function runs once when the program starts. In this function, we set the **pinMode** of each motor pin to OUTPUT, allowing the Arduino to control motor direction and speed:

```cpp

    pinMode(R2, OUTPUT);
    pinMode(pwmR, OUTPUT);

    pinMode(L2, OUTPUT);
    pinMode(pwmL, OUTPUT);
}
```

## Driving the Rover

We control the rover’s direction and speed by adjusting two variables:
- `v`: **Linear velocity** for forward or backward movement.
- `w`: **Angular velocity** for rotation.

By assigning values to these variables, we create velocity vectors for different motions. Here’s an example that drives the rover forward, backward, turns left, and then right:

```cpp
float v = 0.346;  // Full forward speed
float w = 0;      // No rotation
motor_controller(v, w);
delay(2000);

v = -0.346;       // Full backward speed
motor_controller(v, w);
delay(2000);

v = 0;
w = 4.73;         // Full left rotation
motor_controller(v, w);
delay(2000);

w = -4.73;        // Full right rotation
motor_controller(v, w);
delay(2000);
```
