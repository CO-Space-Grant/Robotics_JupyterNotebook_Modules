## Localization: Where is the Rover?

**Localization** refers to the rover’s ability to **estimate its position and orientation** within an environment. This process is essential for **autonomous navigation**, as it allows the robot to know where it is in relation to the world around it. Robots typically use a **combination** of wheel encoders, inertial measurement units (IMUs), and more advanced sensors like LiDAR or visual odometry to collect information about their movements and surroundings. However, basic sensors like wheel encoders **alone** often **accumulate error** due to wheel slippage or terrain irregularities. That’s why modern localization incorporates more advanced **sensor fusion techniques**.


![Localization](Localization.png)

### Why is Localization Important?

Without localization, the rover cannot make **intelligent navigation decisions**. Attempting to use wheel odometry alone is unreliable due to slippage. Referencing Module 7, we can use Sensor Fusion techniques (such as Kalman Filtering) to improve localization accuracy.

### How Does Localization Work?

The rover determines its position by:

1. **Dead Reckoning**: Estimating position using sensor data (IMU, encoders).
2. **Landmark-Based Localization**: Using known obstacles or visual markers.
3. **Sensor Fusion**: Combining IMU, LiDAR/Time-of-Flight (ToF) sensors, and camera data.


## Mapping: How Does the Rover Understand the Environment?

Mapping is the process of **creating a spatial representation** of the environment, often in the form of an **[occupancy grid](https://tse1.mm.bing.net/th/id/OIP.IAdfpuLspMLbzCptHIHozgHaFH?rs=1&pid=ImgDetMain)** or a **[feature map](https://th.bing.com/th/id/OIP.P0sd5EgiLs3RyzB3-orNRgHaCY?w=301&h=180&c=7&r=0&o=5&dpr=1.5&pid=1.7)**. As a robot navigates an area, it uses sensors to **detect features** such as walls, furniture, or obstacles. This information is then stored in a **data structure** that the robot references for **navigation and planning**. There are generally two mapping strategies: grid-based and feature-based. In **grid-based mapping**, the space is divided into **uniform cells**, each of which is marked as free, occupied, or unknown based on sensor data. **Feature-based mapping** instead focuses on identifying and **storing distinct environmental markers**.


![Mapping](Mapping.png)

### Why is Mapping Important?

Mapping is essential because it enables the rover to operate safely and efficiently. Without a map, the robot would not be able to make informed decisions about where to move or how to avoid obstacles. Additionally, if the robot loses track of its location, it can use the known features in its map to re-localize itself. This ability to reorient within the environment based on previously seen landmarks is a key element of SLAM, allowing mapping and localization to inform and support each other.


### How Does Mapping Work?

1. **Grid-Based Mapping**: The environment is divided into cells, where obstacles are marked.
2. **Feature-Based Mapping**: Key environmental features (walls, objects) are recorded.
3. **SLAM (Simultaneous Localization and Mapping)**: The rover localizes and maps at the same time, correcting errors dynamically.

An example of SLAM would be a LiDAR sensor or ToF sensor scanning the environment while the rover moves, creating an obstacle map in real time.

# Simultaneous Localization and Mapping (SLAM) 

SLAM is the integration of the **two tasks** described above: **localizing** the robot while it **constructs a map** of its environment. This presents a unique challenge, as both localization and mapping depend on each other to be accurate. A verifiable Chicken and the Egg paradox. SLAM systems typically rely on **iterative algorithms** that refine the robot’s position as new sensor data arrives, while simultaneously updating the map based on this estimated position. Algorithms such as **EKF-SLAM** or **GraphSLAM** are designed to handle these complex feedback loops.

In practice, a rover equipped with SLAM might use a **LiDAR or ToF sensor** to scan its surroundings while it moves, with the data being used to both update its internal map and correct any drift in its estimated position. The **quality** of a SLAM system is determined by how well it **handles sensor noise, how efficiently it processes data, and how robust it is in dynamic or feature-scarce environments**. Although implementing SLAM on hardware-constrained platforms can be difficult, understanding its principles is essential for designing effective autonomous systems.

SLAM can take **many forms** depending on the sensors used and the environment in which the robot operates. **LiDAR SLAM** uses laser scanners to produce highly accurate 2D or 3D maps, while **Visual SLAM** leverages camera data to extract features and track motion. **Depth SLAM** relies on ranging information, often from Time-of-Flight (ToF) sensors, to calculate depth data. **Visual-only SLAM** uses image sequences without depth sensors, while **Visual-Inertial (VI) SLAM** combines camera inputs with IMU readings for improved robustness. **RGB-D SLAM** utilizes both color and depth information, often from RGB-D cameras, for enhanced environmental understanding. Lastly, **Graph SLAM** represents the mapping problem as a factor graph, optimizing over all past poses and measurements to provide a globally consistent estimate of the map and the robot's trajectory.
